```python
import smtplib
from email.message import EmailMessage

def sendMail(to, subject, body):

    msg = EmailMessage()
    msg['From'] = GMAIL_ID
    msg['To'] = to
    msg['Subject'] = subject
    msg.set_content(body)

    with smtplib.SMTP_SSL('smtp.gmail.com', 465) as server:
        server.login(GMAIL_ID, GMAIL_PASSWORD)
        server.send_message(msg)

    print(f"Mail sent successfully to {to}")
```

```python
import imaplib
import email
from email import policy

def receiveMail(criteria='UNSEEN', limit=10):

    messages = []

    with imaplib.IMAP4_SSL('imap.gmail.com') as server:

        server.login(GMAIL_ID, GMAIL_PASSWORD)
        server.select('INBOX', readonly=True)

        status, data = server.uid('search', None, criteria)

        if status != 'OK':
            return messages

        mail_ids = data[0].split()

        for uid in reversed(mail_ids[-limit:]):

            status, data = server.uid(
                'fetch', uid, '(BODY.PEEK[])'
            )

            if status != 'OK':
                continue

            msg = email.message_from_bytes(
                data[0][1], policy=policy.default
            )

            body = msg.get_body(
                preferencelist=('plain', 'html')
            )

            messages.append({
                'uid': uid.decode(),
                'from': str(msg['From']),
                'to': str(msg['To']),
                'subject': str(msg['Subject']),
                'date': str(msg['Date']),
                'message_id': str(msg['Message-ID']),
                'body': body.get_content() if body else ''
            })

    return messages
```
